# Hello World Validation

This notebook reads the local curated hello-world summary and validates the deterministic expected values.

```python
from pathlib import Path
import json

summary_path = Path('/home/jovyan/data/curated/hello_world/latest/summary.json')
assert summary_path.is_file(), f'Missing curated summary: {summary_path}'
summary = json.loads(summary_path.read_text(encoding='utf-8'))
summary
```

```python
expected = {
    'job_name': 'hello_world',
    'run_date': '2026-04-24',
    'record_count': 4,
    'total_amount': 100,
    'category_counts': {'alpha': 2, 'beta': 1, 'gamma': 1},
    'minimum_date': '2026-01-10',
    'maximum_date': '2026-01-13',
}

for key, value in expected.items():
    assert summary[key] == value, f'Unexpected {key}: {summary[key]!r}'

print('hello_world validation passed')
```
